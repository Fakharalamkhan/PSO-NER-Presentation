## 🏷️ Part-of-Speech (POS) Tagging

<div style="padding: 15px; border-left: 5px solid #0076B6; background-color: #F0F7FB; border-radius: 4px; margin-bottom: 20px;">
   <h4 style="margin-top: 0; color: #0076B6; font-family: sans-serif;"><b>💡 Concept: What is POS Tagging?</b></h4>
   <p style="margin-bottom: 0; font-family: sans-serif; font-size: 14px; line-height: 1.6;">
       <b>Part-of-Speech (POS) Tagging</b> is the process of marking up a word in a text (corpus) as corresponding to a particular part of speech, based on both its definition and its context (i.e., its relationship with adjacent and related words in a phrase, sentence, or paragraph).<br>
       Common parts of speech include: <b>Nouns (nouns, pronouns), Verbs, Adjectives, Adverbs, Prepositions, Conjunctions, and Articles.</b>
   </p>
</div>

<div style="padding: 15px; border-left: 5px solid #28A745; background-color: #F4FBF5; border-radius: 4px; margin-bottom: 20px;">
   <h4 style="margin-top: 0; color: #28A745; font-family: sans-serif;"><b>⚙️ How it Works in spaCy</b></h4>
   <p style="margin-bottom: 0; font-family: sans-serif; font-size: 14px; line-height: 1.6;">
       spaCy performs POS tagging using a statistical model. It uses two layers of tags:
       <ul style="margin-top: 5px; margin-bottom: 0;">
           <li><b>Universal POS (UPOS):</b> A coarse-grained tag set (e.g., <code>NOUN</code>, <code>VERB</code>, <code>ADJ</code>) that is consistent across all languages. Available via <code>token.pos_</code>.</li>
           <li><b>Detailed POS Tag:</b> A fine-grained tag set (e.g., <code>NNP</code> for proper singular noun, <code>VBD</code> for past tense verb) specific to the English Penn Treebank. Available via <code>token.tag_</code>.</li>
       </ul>
   </p>
</div>

```python
# A simple sentence to demonstrate POS Tagging
pos_sentence = "The quick brown fox jumps over the lazy dog."

# Process the sentence using spaCy nlp pipeline
doc_pos = nlp(pos_sentence)

# Extract and print POS tags in a structured table
pos_data = []
for token in doc_pos:
    pos_data.append({
        "Token/Word": token.text,
        "Coarse Tag (UPOS)": token.pos_,
        "Fine Tag (Penn)": token.tag_,
        "Explanation": spacy.explain(token.tag_)
    })

# Convert to DataFrame for beautiful notebook rendering
df_pos = pd.DataFrame(pos_data)
df_pos
```

## 🏢 Named Entity Recognition (NER)

<div style="padding: 15px; border-left: 5px solid #0076B6; background-color: #F0F7FB; border-radius: 4px; margin-bottom: 20px;">
   <h4 style="margin-top: 0; color: #0076B6; font-family: sans-serif;"><b>💡 Concept: What is Named Entity Recognition (NER)?</b></h4>
   <p style="margin-bottom: 0; font-family: sans-serif; font-size: 14px; line-height: 1.6;">
       <b>Named Entity Recognition (NER)</b> (also known as entity identification or entity extraction) is a subtask of information extraction that seeks to locate and classify named entities mentioned in unstructured text into pre-defined categories.<br>
       Common entity categories include:
       <ul style="margin-top: 5px; margin-bottom: 0;">
           <li><b>PERSON:</b> People, including fictional characters.</li>
           <li><b>ORG:</b> Companies, agencies, institutions, organizations.</li>
           <li><b>GPE:</b> Geopolitical entities (countries, cities, states).</li>
           <li><b>DATE:</b> Absolute or relative dates or periods.</li>
           <li><b>MONEY:</b> Monetary values, including unit.</li>
       </ul>
   </p>
</div>

<div style="padding: 15px; border-left: 5px solid #28A745; background-color: #F4FBF5; border-radius: 4px; margin-bottom: 20px;">
   <h4 style="margin-top: 0; color: #28A745; font-family: sans-serif;"><b>⚙️ How it Works: The BIO Tagging Scheme</b></h4>
   <p style="margin-bottom: 0; font-family: sans-serif; font-size: 14px; line-height: 1.6;">
       Because named entities can consist of multiple words (e.g., <i>"New York City"</i> or <i>"Bill Gates"</i>), NLP models use a tagging scheme called <b>BIO tagging</b> to denote entity boundaries:
       <ul style="margin-top: 5px; margin-bottom: 0;">
           <li><b>B-[Entity]:</b> Beginning of a named entity.</li>
           <li><b>I-[Entity]:</b> Inside of a named entity (applies to subsequent words of a multi-word entity).</li>
           <li><b>O:</b> Outside of any named entity.</li>
       </ul>
   </p>
</div>
